# Универсальная система агентов Hello-Agents

## 📝 Описание проекта

Это универсальная система агентов на базе фреймворка **Hello-Agents** с архитектурой **один агент + несколько инструментов**.
Агент регистрирует и вызывает несколько инструментов через ToolRegistry для решения сложных задач.

### Основные функции
- ✅ **Интеллектуальный веб-поиск**: поддержка нескольких поисковых движков и извлечения контента
- ✅ **Безопасное выполнение в терминале**: ограниченное выполнение команд с белым списком
- ✅ **Память**: поддержка запоминания предпочтений и важной информации (в будущем)
- ✅ **Несколько движков**: DuckDuckGo, Brave, Ecosia, Searx

## 🧰 Часть 3: Определение инструментов

В этом проекте используются предопределённые инструменты из каталога `tools/`:

```python
# Импорт определений инструментов
from src.tools.browser_tool import BrowserTool
from src.tools.terminal_tool import TerminalTool

# Создание экземпляров инструментов
browser_tool = BrowserTool()
terminal_tool = TerminalTool()

print("📋 Список доступных инструментов:")
print(f"1. {browser_tool.name}: {browser_tool.description}")
print(f"2. {terminal_tool.name}: {terminal_tool.description}")
```

## 🤖 Часть 4: Создание агента

```python
# Импорт класса агента
from src.agents.agent_universal import UniversalAgent

# Создание экземпляра агента
agent = UniversalAgent()

print("🚀 Агент успешно создан!")
print(f"📛 Имя агента: {agent.name}")
print(f"🔧 Количество зарегистрированных инструментов: {len(agent.tool_registry.tools)}")

# Отображение зарегистрированных инструментов
print("\n📋 Зарегистрированные инструменты:")
for tool_name, tool in agent.tool_registry.tools.items():
    print(f"  - {tool_name}: {tool.description}")
```

## 🎯 Часть 5: Демонстрация функций

### Пример 1: базовый поиск

```python
# Демонстрация поиска
query = "Последние достижения в области ИИ на Python"
print(f"🔍 Поисковый запрос: {query}")
print("=" * 50)

try:
    result = agent.run(query)
    print("\n📋 Результаты поиска:")
    print(result)
except Exception as e:
    print(f"❌ Ошибка поиска: {e}")
```

### Пример 2: выполнение команд в терминале

```python
# Демонстрация терминала (безопасная команда)
command = "pwd"
print(f"💻 Выполняемая команда: {command}")
print("=" * 50)

try:
    result = agent.run(f"Выполни {command}")
    print("\n📋 Результат выполнения:")
    print(result)
except Exception as e:
    print(f"❌ Ошибка выполнения команды: {e}")
```

### Пример 3: обработка сложной задачи

```python
# Демонстрация сложной задачи
complex_task = "Найди информацию о последней версии фреймворка LangChain, затем покажи список файлов в текущем каталоге"
print(f"🎯 Сложная задача: {complex_task}")
print("=" * 50)

try:
    result = agent.run(complex_task)
    print("\n📋 Результат обработки:")
    print(result)
except Exception as e:
    print(f"❌ Ошибка обработки задачи: {e}")
```

## 📊 Часть 6: Оценка производительности (опционально)

```python
import time

# Простой тест производительности
test_queries = [
    "Основы синтаксиса Python",
    "Список файлов текущего каталога",
    "Тенденции развития искусственного интеллекта"
]

print("📊 Начало теста производительности...")
print("=" * 50)

for i, query in enumerate(test_queries, 1):
    print(f"\n🧪 Тест {i}: {query}")
    
    start_time = time.time()
    try:
        result = agent.run(query)
        end_time = time.time()
        
        response_time = end_time - start_time
        result_length = len(result) if result else 0
        
        print(f"✅ Успех — время отклика: {response_time:.2f} с, длина результата: {result_length} символов")
        print(f"📝 Превью результата: {result[:100]}..." if len(result) > 100 else f"📝 Результат: {result}")
    except Exception as e:
        end_time = time.time()
        response_time = end_time - start_time
        print(f"❌ Ошибка — время отклика: {response_time:.2f} с, ошибка: {e}")

print("\n📊 Тест производительности завершён")
```
