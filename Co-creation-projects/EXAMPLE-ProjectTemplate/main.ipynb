## Часть 2: Определение инструментов

```python
class CustomTool(BaseTool):
    """Пользовательский инструмент"""
    
    name = "tool_name"
    description = "Описание инструмента"
    
    def run(self, query: str) -> str:
        """Логика выполнения инструмента"""
        # Реализуйте логику вашего инструмента
        return f"Результат обработки: {query}"

print("✅ Инструмент определён")
```

## Часть 3: Создание агента

```python
# Создание LLM
llm = HelloAgentsLLM()

# Системный промпт
system_prompt = """Вы — интеллектуальный помощник.

Ваши задачи:
1. Понять потребности пользователя
2. Использовать подходящие инструменты
3. Давать полезные ответы
"""

# Создание агента
agent = SimpleAgent(
    name="Пример агента",
    llm=llm,
    system_prompt=system_prompt
)

# Добавление инструмента
agent.add_tool(CustomTool())

print("✅ Агент создан")
```

## Часть 4: Демонстрация функций

```python
# Пример 1: базовая функциональность
print("=== Пример 1: базовая функциональность ===")
result = agent.run("ваш тестовый ввод")
print(result)
```

```python
# Пример 2: сложный сценарий
print("\n=== Пример 2: сложный сценарий ===")
result = agent.run("более сложный тестовый ввод")
print(result)
```
