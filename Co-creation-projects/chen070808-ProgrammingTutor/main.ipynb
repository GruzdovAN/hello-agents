# Интеллектуальный наставник по программированию (Intelligent Programming Tutor)

Персонализированная система обучения программированию на основе мультиагентного взаимодействия, демонстрирующая, как построить сложную систему Agent-to-Agent (A2A) с фреймворком `hello-agents`.

## Архитектура системы

Система использует иерархическую архитектуру агентов:

- **Tutor (наставник)**: главный координирующий агент, взаимодействует с пользователем и вызывает подагентов
- **Planner (планировщик)**: анализирует потребности пользователя и составляет персональный план обучения
- **Exercise (составитель задач)**: генерирует целевые упражнения по программированию на основе учебного материала
- **Reviewer (рецензент)**: проверяет код пользователя и даёт профессиональную обратную связь

## Технические особенности

1. **Мультиагентное взаимодействие**: подагенты обёрнуты в `AgentTool` для A2A-коммуникации
2. **Вызов инструментов**: у Reviewer есть инструмент `CodeRunner` для выполнения Python-кода
3. **Модульный дизайн**: у каждого агента одна зона ответственности, легко поддерживать и расширять

## О демонстрации

Этот notebook содержит три полных тестовых сценария, демонстрирующих ключевые функции интеллектуального наставника.


## Шаг 1: Настройка окружения

Инициализация LLM и конфигурации окружения.

**Ключевые компоненты**:
- `HelloAgentsLLM`: единый интерфейс LLM
- файл `.env`: хранение API-ключей и других секретов
- путь `src`: пользовательские агенты и инструменты


````python
# 1. Настройка окружения
import os
import sys
from dotenv import load_dotenv
from hello_agents import HelloAgentsLLM

load_dotenv()

if "src" not in sys.path:
    sys.path.append(os.path.abspath("src"))

# Инициализация LLM
llm = HelloAgentsLLM()

print("✅ Окружение настроено")
print("✅ LLM инициализирован")
````

## Шаг 2: Инициализация интеллектуального наставника

При создании экземпляра `TutorAgent` автоматически:
1. Создаются три подагента: Planner, Exercise, Reviewer
2. Подагенты оборачиваются в инструменты (`call_planner`, `call_exercise`, `call_reviewer`)
3. Инструмент `CodeRunner` регистрируется для Reviewer

**Архитектурные особенности**:
- `AgentTool` реализует вызов Agent-to-Agent
- У каждого подагента свой `system_prompt`, определяющий область экспертизы
- Tutor координирует всех подагентов через вызов инструментов


````python
# 2. Инициализация Tutor (автоматически создаёт всех подагентов)
from agents.tutor import TutorAgent

print("Создание интеллектуального наставника...")
tutor = TutorAgent(llm)

print("\n✅ Tutor инициализирован!")
print("   - Planner (планировщик) готов")
print("   - Exercise (составитель задач) готов") 
print("   - Reviewer (рецензент) готов")
````

---

## Тест 1: запрос плана обучения

Демонстрация взаимодействия **Tutor → Planner**.

**Порядок выполнения**:
1. Пользователь сообщает Tutor свою учебную цель
2. Tutor распознаёт намерение и вызывает инструмент `call_planner`
3. Planner анализирует запрос и формирует модульный план обучения
4. Tutor дружелюбно представляет план пользователю

**Ожидаемый результат**: полный учебный маршрут с модулями, расписанием и рекомендациями.


````python
user_goal = "Я хочу изучить list comprehensions в Python"
print(f"Цель пользователя: {user_goal}\n")

# Tutor вызовет инструмент call_planner
response = tutor.run(f"Пользователь сказал: '{user_goal}'. Составь для него план обучения.")

print("=== Ответ Tutor ===")
print(response)
````

---

## Тест 2: запрос упражнения

Демонстрация взаимодействия **Tutor → Exercise**.

**Порядок выполнения**:
1. Пользователь просит Tutor упражнение
2. Tutor вызывает инструмент `call_exercise`
3. Exercise генерирует структурированное упражнение по программированию
4. Tutor возвращает полное задание с описанием, примерами и ограничениями

**Ожидаемый результат**: качественное упражнение, включающее:
- описание задачи
- примеры ввода/вывода
- ограничения
- сигнатуру функции


````python
# Tutor вызовет инструмент call_exercise
response = tutor.run("Дай мне упражнение по list comprehensions.")

print("=== Ответ Tutor ===")
print(response)
````

---

## Тест 3: отправка кода на ревью

Демонстрация многоуровневого взаимодействия **Tutor → Reviewer → CodeRunner**.

**Порядок выполнения**:
1. Пользователь отправляет код Tutor на проверку
2. Tutor вызывает инструмент `call_reviewer`
3. Reviewer использует инструмент `code_runner` для выполнения и проверки кода
4. Reviewer анализирует логику, стиль и производительность
5. Tutor возвращает подробный отчёт о ревью

**Технические особенности**:
- Reviewer может реально запускать код и получать вывод
- Структурированная обратная связь (плюсы, проблемы, рекомендации)
- Поддержка фрагментов кода и определений функций

**О тестовом коде**:
Пользователь пытается решить упражнение из теста 2 (отфильтровать положительные чётные числа и возвести в квадрат); Reviewer проверит корректность.


````python
# Пользователь пытается решить упражнение выше
user_code = """
def filter_and_square_numbers(numbers):
    return [n * n for n in numbers if n > 0 and n % 2 == 0]

# Тест
test_list = [1, 2, 3, 4, 5, 6, -2, -4, 0]
result = filter_and_square_numbers(test_list)
print(result)
"""

print(f"=== Код пользователя ===\n{user_code}\n")

# Tutor вызовет инструмент call_reviewer
response = tutor.run(f"""Пользователь пытается решить предыдущее упражнение по list comprehensions. Проверь следующий код:

{user_code}

Требование задачи: отфильтровать положительные чётные числа и вернуть их квадраты.""")

print("=== Ответ Tutor ===")
print(response)
````

## Итоги

Эта демонстрация показывает, как построить мультиагентную систему на фреймворке `hello-agents`.

### Ключевые технологии

1. **AgentTool**: оборачивает агента в инструмент для A2A-вызовов
   ```python
   self.add_tool(AgentTool(
       self.planner,
       name="call_planner",
       description="Вызов планировщика курса"
   ))
   ```

2. **Цепочка инструментов**: Reviewer использует CodeRunner для выполнения кода
   ```python
   ReviewerAgent(llm, tools=[CodeRunner()])
   ```

3. **System Prompt**: тщательно спроектированные промпты определяют поведение агента
